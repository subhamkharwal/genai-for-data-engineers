```python
# Load GPT-4's tokenizer (cl100k_base encoding)
from tiktoken import get_encoding
enc = get_encoding("cl100k_base")
```

```python
# Max token value
enc.max_token_value
```

```python
# Tokenize text
_input = "My name is Subham"
token_ids = enc.encode(_input)
print(token_ids)
```

```python
# Decode token ids
enc.decode_single_token_bytes(token_ids[4])
enc.decode(token_ids)
```

```python
# For diffent text same meaning -> different token size
same_sentence = {
    "English": "Hello, my name is Alice and I love learning.",
    "Spanish": "Hola, mi nombre es Alicia y me encanta aprender.",
    "Hindi":   "नमस्ते, मेरा नाम ऐलिस है और मुझे सीखना पसंद है।",
    "Chinese":  "你好，我叫爱丽丝，我喜欢学习。",
    "Arabic":  "مرحبا، اسمي أليس وأحب التعلم.",
}

print("=== Token count by language (same meaning) ===")
for language, sentence in same_sentence.items():
    count = len(enc.encode(sentence))
    print(f"  {language:10s}: {count:3d} tokens — {sentence}")
print()
```
